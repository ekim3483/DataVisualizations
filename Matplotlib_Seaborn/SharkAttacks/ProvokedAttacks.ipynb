```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_sharkdata2.csv')
```

```python
# While skimming the dataset, my curiosity was piqued by the 'provoked' column.
```

```python
df_provoked = df[df['type'] == 'Provoked']
```

```python
m_count = df_provoked['sex_'].value_counts()['M']
```

```python
f_count = df_provoked['sex_'].value_counts()['F']
```

```python
!pip install pywaffle -q
import matplotlib.pyplot as plt
from pywaffle import Waffle
```

```python
LuckyIdiotsAndDarwinAward = {'Male': m_count, 'Female': f_count}

plt.figure(
  FigureClass=Waffle,
  rows=5,
  values=LuckyIdiotsAndDarwinAward,
  icons='star',
  colors=['#99ccff','#ffc0cb'],
  figsize=(8, 4), 
  font_size = 40,
  legend={
        'loc': 'lower left',
        'bbox_to_anchor': (0, -0.15),
        'ncol': len(LuckyIdiotsAndDarwinAward),
        'fontsize': 12
    }
)

plt.style.use('dark_background')
plt.title('People who provoked sharks and were thus injured or killed:')
plt.show()
```
